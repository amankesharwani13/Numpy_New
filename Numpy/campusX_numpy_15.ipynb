```python
# getting the dimensions
a1 = np.array([1, 2, 4, 5, 6])
a2 = np.arange(0, 10).reshape(5, 2)
a3 = np.arange(0, 8).reshape(2, 2, 2)
print(a1.ndim)
print(a2.ndim)  
print(a3.ndim)
```

```python
# Array Functions
a4 = np.random.random((3, 3))
a4 = np.round(a4*100)
a4   # genarating random 3x3 matrix with values between 0 to 100
```

```python
# max/min/sum/prod
np.sum(a4)
```

```python
# finding max row or column wise
print(np.max(a4, axis=0))  # column wise
print(np.max(a4, axis=1))  # row wise
```

```python
# mean/mediam/std/var
print(np.mean(a4))
print(np.median(a4))
print(np.std(a4))
print(np.var(a4))
```

```python
# trignometric functions
np.sin(a4)
```

```python
# dot product
a5 = np.arange(0, 12).reshape(3, 4)
a6 = np.arange(12, 24).reshape(4, 3)
np.dot(a5, a6)
```

```python
# log and exponential
print(np.log(a5))
print(" ")
print(np.exp(a5))
```

```python
# round/floor/ceil
a = np.random.random((2, 3))*100
print(a)
print("Rounded Values:")
print(np.round(a))
print("Floor Values:")
print(np.floor(a))
print("Ceil Values:")
print(np.ceil(a))
```

```python
print(a1)
print("2D")
print(a2)
print("3D")
print(a3)
```

```python
# Indexing and Slicing
print(a1[-1])
print(a2[3, 1])  # accessing element at 4th row and 2nd column
print(a3[1, 0, 1])  # accessing element at 2nd block, 1st row and 2nd column
```

```python
# slicing
print(a1[1:4])
print(a2[:,1])
print(a2[1,:])
print(" ")
print(a2[1:3, 0:2])
print(" ")
print(a3[:1, 1:2])
```

```python
# Iterating Arrays
for i in a1:
    print(i)
```

```python
for i in a2:
    for j in i:
        print(j)
```

```python
for i in np.nditer(a3):
    print(i)
```

```python
# transpose
np.transpose(a2)
```

```python
# ravel -> convert n-dim array to 1-dim array
np.ravel(a3)
```

```python
# horizontal and vertical stacking
a7 = np.arange(0, 12).reshape(3,4)
a8 = np.arange(12, 24).reshape(3,4)
print(np.hstack((a7, a8)))  ## horizontal stacking
print(" ")
np.vstack((a7, a8))  ## vertical stacking
```
